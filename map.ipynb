```python
import geopandas as gpd
import numpy as np
import plotly.express as px

gdf_trentino = gpd.read_file(f'./ComuniTrentini.geojson')
gdf_trentino['random_value'] = np.random.rand(len(gdf_trentino))

fig = px.choropleth_map(
    gdf_trentino,
    geojson= gdf_trentino,
    locations=gdf_trentino.index,
    color_continuous_scale='Blues_r',
    color='random_value',
    center={'lat': 46, 'lon': 11},
    zoom=7
)

fig.show()
```
