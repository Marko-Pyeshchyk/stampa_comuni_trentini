```python
import geopandas as gpd
import numpy as np
import plotly.express as px
import plotly.graph_objects as go

gdf_trentino = gpd.read_file(f'./ComuniTrentini.geojson')
gdf_trentino['random_value'] = np.random.rand(len(gdf_trentino))

fig = px.choropleth_map(
    gdf_trentino,
    geojson= gdf_trentino,
    locations=gdf_trentino.index,
    color_continuous_scale='Reds_r',
    color='random_value',
    center={'lat': 46, 'lon': 11},
    zoom=7
)

fig.show()
```

```python
fig = go.Figure(go.Choroplethmap(
    geojson=gdf_trentino.__geo_interface__,
    locations=gdf_trentino.index,
    z=gdf_trentino['random_value'],
    colorscale='Reds_r',
    marker_opacity=0.7,
    marker_line_width=0.5,
))

fig.update_layout(
    map_center={'lat': 46, 'lon': 11},
    map_zoom=7,
)

fig.show()
```
